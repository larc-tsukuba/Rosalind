## 問題

重み付き文字列（weighted string）の **prefix spectrum（接頭辞スペクトル）** とは、その文字列のすべての接頭辞（prefix）の **重み（weight）** を集めたものです。

### 入力

\(n\) 個（\(n \le 100\)）の正の実数からなるリスト \(L\)。

### 出力

prefix spectrum が \(L\) と等しくなる、長さ \(n-1\) のタンパク質配列を1つ出力してください。

複数の解が存在する場合は、そのうちどれを出力しても構いません。

アミノ酸の質量については、**monoisotopic mass table（モノアイソトピック質量表）**を参照してください。


### 補足説明
## 質量分析入門

「Calculating Protein Mass（タンパク質質量の計算）」では、**質量分析（mass spectrometry）** と呼ばれる分析化学的手法について簡単に触れました。質量分析は、粒子や分子の **質量電荷比（mass-to-charge ratio）** を測定することを目的としています。

質量分析計では、まず試料を蒸発させて気体にし、その後、試料中の粒子を **イオン化（ionized）** します。生成したイオンを電磁場に置くと、イオンはその電荷と質量に応じて分離されます。

質量分析計から得られる結果は **質量スペクトル（mass spectrum）** と呼ばれます。これは、イオンが取り得る質量電荷比の値と、それぞれの質量電荷比をもつイオンの **強度（intensity）**、すなわち実際に観測された頻度を示したグラフです。

ここではひとまず電荷については無視し、イオンの **モノアイソトピック質量（monoisotopic mass）** の一覧を **簡略化スペクトル（simplified spectrum）** として考えることにします。

研究者は、タンパク質をアミノ酸1個ずつ直接観察できるような安価な技術を持っているわけではありません。分子は非常に小さすぎるためです。その代わり、タンパク質の構造を決定するために、同じタンパク質を複数コピー用意して、それぞれをより小さな断片に分割し、得られた断片の質量を測定します。

このとき、それぞれの **切断（cut）**、つまり切れる位置は、2つの **アミノ酸** の間で起こるものと仮定します。また、考えられるすべての切断について、生成する断片の質量を測定できるものとします。

例えば、未知のタンパク質

`PRTEIN`

は、次の5通りの位置で切断できます。

`P` と `RTEIN`
`PR` と `TEIN`
`PRT` と `EIN`
`PRTE` と `IN`
`PRTEI` と `N`

このようにして、タンパク質全体を含むすべての断片の質量を測定できます。

タンパク質の「左端」は **N末端（N-terminus）** と呼ばれます。タンパク質配列の **prefix（接頭部分）**

`P`, `PR`, `PRT`, `PRTE`, `PRTEI`

に対応するイオンは、**b-ion（bイオン）** と呼ばれます。

一方、タンパク質の「右端」は **C末端（C-terminus）** と呼ばれます。タンパク質配列の **suffix（接尾部分）**

`N`, `IN`, `EIN`, `TEIN`, `RTEIN`

に対応するイオンは、**y-ion（yイオン）** と呼ばれます。

隣り合う2つの b-ion、あるいは y-ion の質量の差を取ると、1個のアミノ酸 **残基（residue）** の質量が得られます。

例えば、

`PRT`

と

`PR`

の質量の差は、必ずアミノ酸 `T` の質量になります。

したがって、この考え方を拡張すると、あるタンパク質について **すべての b-ion の質量が分かっていれば、そのタンパク質のアミノ酸配列を復元することができます。**


```python
# リンクから取得
mass_table = {
    "A": 71.03711,
    "C": 103.00919,
    "D": 115.02694,
    "E": 129.04259,
    "F": 147.06841,
    "G": 57.02146,
    "H": 137.05891,
    "I": 113.08406,
    "K": 128.09496,
    "L": 113.08406,
    "M": 131.04049,
    "N": 114.04293,
    "P": 97.05276,
    "Q": 128.05858,
    "R": 156.10111,
    "S": 87.03203,
    "T": 101.04768,
    "V": 99.06841,
    "W": 186.07931,
    "Y": 163.06333
}
```

```python
mass_input = """3524.8542
3710.9335
3841.974
3970.0326
4057.0646"""
```

```python
def parse_mass_input(mass_input):
    return list(map(float, mass_input.splitlines()))



def get_mass_difference(mass_input):
    mass_difference = []
    for i, j in zip(mass_input, mass_input[1:]):
        mass_difference.append(round(j - i, 5))
    return mass_difference


def get_protein_sequence(mass_difference, mass_table):
    protein_sequence = []
    for mass in mass_difference:
        for amino_acid, amino_acid_mass in mass_table.items():
            if round(amino_acid_mass, 2) == round(mass, 2):
                protein_sequence.append(amino_acid)
                break
    return ''.join(protein_sequence)
```

```python
get_mass_difference(parse_mass_input(mass_input))
```

```python
get_protein_sequence(get_mass_difference(parse_mass_input(mass_input)), mass_table)
```
