## 説明の外注


[トライ木(Trie木) の解説と実装【接頭辞(prefix) を利用したデータ構造】](https://algo-logic.info/trie-tree/)


## 問題の入出力

- 入力: 最大100本、各配列の長さが最大100 bpのDNA配列

```
ATAGA
ATC
GAT
```

- 出力: 構築したTrie木の隣接リスト
    - 親ノード
    - 子ノード
    - 塩基

```
1 2 A
2 3 T
3 4 A
4 5 G
5 6 A
3 7 C
1 8 G
8 9 A
9 10 T
```

## 実装


```text
1
└── A → 2
         └── T → 3
                  ├── G → 4
                  └── C → 5
```

といったTrie木を、どう格納するか。

自然言語では以下を実装すれば良さそう：

- 「ノード1にAが来たら、次のノード2に行く」
- 「ノード2にTが来たら、次のノード3に行く」
- 「ノード3にGが来たら、次のノード4に行く」
- 「ノード3にCが来たら、次のノード5に行く」

Pythonでは辞書を使えば、以下のように表現できる：

```python
trie = {
    (1, "A"): 2,
    (2, "T"): 3,
    (3, "G"): 4,
    (3, "C"): 5
}
```



## サンプル

````python
from pathlib import Path
sequenses = Path("sample.txt").read_text().splitlines()

print(sequenses)

trie = {}
next_node = 2 # ← 次に新しく作成するノード番号

for seq in sequenses:
    current_node = 1
    for nucreotide in seq:
        # 現在のノードと塩基をペアにする
        node_nucreotide = (current_node, nucreotide)

        # 初見のペアの場合、trie木に登録して、
        # そのvalueに次の行き先となるノードの番号を当てる
        if node_nucreotide not in trie:
            trie[node_nucreotide] = next_node
            next_node += 1

        # current_nodeを次の行き先へと更新する
        current_node = trie[node_nucreotide]

for (current_node, nucreotide), next_node in trie.items():
    print(current_node, next_node, nucreotide)
````

## 本番

````python
from pathlib import Path
sequenses = Path("rosalind_trie.txt").read_text().splitlines()

trie = {}
next_node = 2
for seq in sequenses:
    current_node = 1
    for nucreotide in seq:
        node_nucreotide = (current_node, nucreotide)
        
        if node_nucreotide not in trie:
            trie[node_nucreotide] = next_node
            next_node += 1

        current_node = trie[node_nucreotide]

ans = []
for (current_node, nucreotide), next_node in trie.items():
    a = " ".join([str(current_node), str(next_node), nucreotide])
    ans.append(a)

Path("answer.txt").write_text("\n".join(ans))
````
