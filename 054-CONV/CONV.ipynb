# 054-CONV

```python
A = [186.07931, 287.12699, 548.20532, 580.18077, 681.22845, 706.27446, 782.27613, 968.35544, 968.35544]
B = [101.04768, 158.06914, 202.09536, 318.09979, 419.14747, 463.17369]
```

```python
from collections import Counter

ans = Counter()

for a in A:
    for b in B:
        ans[round(a - b, 5)] += 1

x, count = ans.most_common(1)[0]

print(count)
print(abs(x))
```

```python
from pathlib import Path
sample = Path("rosalind_conv.txt").read_text().strip().splitlines()
A = [float(x) for x in sample[0].split()]
B = [float(x) for x in sample[1].split()]
```

```python
from collections import Counter

ans = Counter()

for a in A:
    for b in B:
        ans[round(a - b, 5)] += 1

x, count = ans.most_common(1)[0]

print(count)
print(abs(x))
```
